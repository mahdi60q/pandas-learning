# Filtering Data with Pandas Query

This notebook demonstrates how to use the Pandas `query()` method to filter and analyze DataFrame data.

## Topics Covered

- Filtering rows with `query()`
- Using comparison operators
- Combining conditions
- Querying string and numeric columns
- Using query results in data analysis

## Goal

The goal is to learn how `query()` can make DataFrame filtering more readable and expressive.

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
```

# Query

```python
df = pd.read_csv('../data/stocks.csv')
df
```

```python
df[df['Symbol'] == 'AAPL']
```

```python
df[df.Symbol == 'AAPL']
```

```python
df.query("Symbol == 'AAPL'")
```

```python
set(df['Symbol'])
```

```python
df['Symbol'].value_counts()
```

```python
df.groupby('Volume').mean(numeric_only=True)
```

```python
group = df.groupby('Symbol')
```

```python
for symbol , data , in group:
    print(symbol)
    print(data)
```

```python
df_groups = pd.concat([data for symbol , data in df.groupby('Symbol')])
```

```python
df_groups
```

```python
df.groupby('Symbol')[['Close' , 'Volume']].mean()
```

```python
df.groupby('Symbol')[['Close']].mean()
```

```python
temp = df.groupby('Symbol')[['Close' , 'Volume']].mean()
temp[temp.Close < 100]
```

```python
df.groupby('Symbol')[['Close' , 'Volume']].mean().query('Close < 100')
```
