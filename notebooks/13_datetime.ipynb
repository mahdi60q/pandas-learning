# DateTime in Pandas

```python
date_sr =pd.Series(pd.date_range('2022-11-15' , periods=3 , freq='D'))
date_sr
```

```python
list = ['Day1','Day2','Day3']

date_sr.index = list
```

```python
date_sr
```

```python
change_format = date_sr.dt.strftime('%d,%m,%Y') 
print(change_format)
```

```python
date_sr = pd.to_datetime(pd.Series('2022-11-15'))
date_sr
```

```python
change_format = date_sr.dt.strftime('%d/%m/%Y')
print(change_format)
```

```python
df = pd.DataFrame({'date': ['3/10/2022', '3/11/2022', '3/12/2022'],
                   'value': [100, 200, 300]})
df['date'] = pd.to_datetime(df['date'])
df
```

```python
df = pd.DataFrame({'date': ['3/10/2022', '3/11/2022', '3/12/2022'],
                   'value': [100, 200, 300]})

df['date'] = pd.to_datetime(df['date'], dayfirst=True)
df
```

```python
df = pd.DataFrame({'date': ['2022-6-10 20:30:0', 
                            '2022-7-1 19:45:30', 
                            '2022-10-12 4:5:1'],
                   'value': [100, 200, 300]})
df['date'] = pd.to_datetime(df['date'], format="%Y-%d-%m %H:%M:%S")
df
```

```python
df = pd.DataFrame({'year': [2022, 2023],
                   'month': [2, 3],
                   'day': [4, 5]})
df['date'] = pd.to_datetime(df)
df
```

```python
df = pd.DataFrame({'name': ['Soheil', 'Mahtab', 'Baran'],
                 'DoB': ['10-09-1989', '06-28-2020', '06-28-2022']})
df['DoB'] = pd.to_datetime(df['DoB'])
```

```python
df['year']= df['DoB'].dt.year
df['month']= df['DoB'].dt.month
df['day']= df['DoB'].dt.day
df
```

```python
dw_mapping={
    0: 'Saturday',
    1: 'Sunday',
    2: 'Monday', 
    3: 'Tuesday', 
    4: 'Wednesday', 
    5: 'Thursday', 
    6: 'Friday'
} 
df['day_of_week_name']=df['DoB'].dt.weekday.map(dw_mapping)
df
```

```python
dw_mapping={
    0: 'شنبه',
    1: 'یکشنبه',
    2: 'دوشنبه', 
    3: 'سه شنبه', 
    4: 'چهارشنبه', 
    5: 'پنجشنبه', 
    6: 'جمعه',
} 
df['روز هفته']=df['DoB'].dt.weekday.map(dw_mapping)
df
```
