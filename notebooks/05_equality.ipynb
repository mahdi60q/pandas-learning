# How to Check Equality  

```python
df = pd.DataFrame({'First' : [1,2,np.nan,3,4] , 'Second' : [1,2,np.nan,3,4]})
```

```python
df
```

```python
df.First == df.Second
```

```python
np.nan == np.nan
```

```python
df.First.equals(df.Second)
```

```python
df_new = df.copy()
df_new.equals(df)
```

```python
df.isna()
```

```python
df.isin([4]) # find number 4
```

```python
df.value_counts()
```

```python
df.isnull().value_counts()
```

```python
df.isetitem(1 , 5)
```
