# Exploding Columns in Pandas

This notebook demonstrates how to transform list-like values stored inside DataFrame cells into separate rows.

## Topics Covered

- Using `explode()`
- Expanding list-like values
- Creating multiple rows from a single row
- Working with nested list data
- Handling exploded DataFrames

## Goal

The goal is to understand how `explode()` can transform nested or list-like data into a tabular structure.

```python
import pandas as pd
```

```python
d = {"Team":["FC Barcelona", "FC Real Madrid"], 
    "Players":[["Ter Stegen", "Semedo", "Piqué", "Lenglet", "Alba", "Rakitic", "De Jong", "Sergi Roberto", "Messi", "Suárez", "Griezmann"], \
               ["Courtois", "Carvajal", "Varane", "Sergio Ramos", "Mendy", "Kroos", "Valverde", "Casemiro", "Isco", "Benzema", "Bale"]]}
```

```python
df = pd.DataFrame(d)
```

```python
df
```

```python
df_player = df.explode("Players")
```

```python
df_player
```

```python
df_player.index
```

```python
df_team = df.explode("Team")
```
