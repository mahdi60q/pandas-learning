# Pandas Data Conversion

This notebook demonstrates different techniques for converting and transforming data using Pandas.

## Topics Covered

- Mapping values with `map()`
- Converting categorical data
- Using `factorize()`
- Converting continuous values
- Creating bins with `cut()`
- Creating quantile-based bins with `qcut()`
- Working with categorical representations

## Goal

The goal is to understand how to convert raw data into useful formats for analysis and machine learning.

```python
import pandas as pd
import numpy as np
```

```python
df_convert = pd.read_csv('../data/titanic.csv')
df_convert.head()
```

```python
pd.DataFrame(df_convert['Sex']).head(5)
```

```python
df_convert['sex_num'] = df_convert['Sex'].map({'male' : 0 , 'female' : 1})
```

```python
df_convert['sex_num']
```

```python
df_convert['sex_factorize'] = df_convert['Sex'].factorize()[0]
```

```python
df_convert['sex_factorize']
```

```python
df_convert['Sex'].factorize()[1]
```

```python
set(df_convert['Sex'])
```

```python
df_convert.head()
```

```python
pd.DataFrame(pd.cut(df_convert['Age'] , bins=[0 , 18 , 25 , 99] , labels=['child' , 'young' , 'old']))
```

```python
pd.qcut(df_convert['Age'] , q=3)
```

```python
pd.qcut(df_convert['Age'] , q=3).value_counts()
```
