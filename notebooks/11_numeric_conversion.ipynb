# Numeric Data Conversion in Pandas

This notebook demonstrates different techniques for converting data into numeric types.

## Topics Covered

- Converting values with `astype()`
- Using `pd.to_numeric()`
- Handling invalid numeric values
- Working with numeric and non-numeric data
- Comparing different conversion approaches

## Goal

The goal is to understand how to safely convert data into numeric formats for analysis and further processing.

```python
import pandas as pd
```

```python
df = pd.DataFrame({'Column_one':['1.1', '2.2', '3.3'],
                   'Column_two':['4.4', '5.5', '6.6'],
                   'Column_three':['7.7', '8.8', '-']})
df
```

```python
df.dtypes
```

```python
convert = df.astype({'Column_one' : 'float' , 'Column_two' : 'float'}).dtypes
convert
```

```python
convert = df.astype({'Column_one' : 'float' , 'Column_two' : 'float'})
convert
```

```python
df['Column_three'] = pd.to_numeric(df['Column_three'], errors='coerce')
print(df)
```

```python
df.dtypes
```

```python
df['Column_three'] = pd.to_numeric(df.Column_three , errors='coerce').fillna(0)
```

```python
df
```

### Alternative Solution

```python
df = df.apply(pd.to_numeric , errors='coerce').fillna(0)
```
