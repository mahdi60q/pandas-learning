```python
pd.DataFrame(np.random.rand(5 , 10) , columns=list('ABCDEFGHIJ'))
```

```python
B = pd.DataFrame({'column one' : [10 , 30] , 'column two' : [20 , 40]})
B
```

```python
B = B.rename({'column one' : 'test 1 ' , 'column two' : 'test 2 '} , axis='columns')
B
```

```python
B.columns = ['one_1' , 'two_2']
B
```

```python
B.columns = B.columns.str.replace('_' , '')
```

```python
B
```

```python
B.add_prefix('_X_')
```
