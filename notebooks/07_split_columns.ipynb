# Splitting Columns in Pandas

This notebook demonstrates techniques for splitting a single DataFrame column into multiple columns.

## Topics Covered

- Splitting string columns
- Using `str.split()`
- Separating values into multiple columns
- Handling structured string data
- Converting split values into DataFrame columns

## Goal

The goal is to learn how to transform columns containing multiple pieces of information into structured DataFrame columns.

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
```

# Split Columns

```python
df = pd.DataFrame({'Name' : ['mahdi ghorbani' , 'MAHDI GHORBANI'] , 'Location' : ['iran' , 'iran']})
df
```

```python
df[['Firstname' , 'Lastname']] = df.Name.str.split(' ' , expand=True)
df
```

```python
df
```

```python
df['Firstname'] , df['Lastname'] = zip(*df.Name.str.split(' '))
```
