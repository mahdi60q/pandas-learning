# Zip, Tuples and DataFrame Construction

This notebook demonstrates how Python `zip()` and tuples can be used to construct and manipulate Pandas DataFrames.

## Topics Covered

- Using `zip()`
- Combining multiple sequences
- Working with tuples
- Creating lists of tuples
- Converting tuples into DataFrames
- Structuring data before DataFrame creation

## Goal

The goal is to understand how Python data structures can be combined with Pandas to create structured DataFrames.

```python
import pandas as pd
```

# zip 

```python
names = ['ali', 'sara', 'mahdi']
ages = [25, 30, 22]
cities = ['tehran', 'shiraz', 'mashhad']

result = zip(names, ages, cities)
print(list(result))
```

```python
names = ['ali', 'sara', 'mahdi']
ages = [25, 30, 22]

for name, age in zip(names, ages):
    print(f"{name} is {age} years old")
```

```python
pairs = [('ali', 25), ('sara', 30), ('mahdi', 22)]

names, ages = zip(*pairs)
print(names)
print(ages)
```

# Convert lists of tuples into pandas Dataframe

```python
data = [('ali', 25, 'tehran'), ('sara', 30, 'shiraz'), ('mahdi', 22, 'mashhad')]

df = pd.DataFrame(data, columns=['Name', 'Age', 'City'])
print(df)
```

```python
data = [('ali', 25, 'tehran'), ('sara', 30), ('mahdi', 22, 'mashhad', 'engineer')]

df = pd.DataFrame(data, columns=['Name', 'Age', 'City', 'Job'])
print(df)
```
