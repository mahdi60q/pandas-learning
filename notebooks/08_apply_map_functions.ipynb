# Apply, Map, Applymap  

```python
df = pd.read_csv('../data/employee.csv')
df.head()
```

```python
set(df['Department'])
```

```python
df.index
```

```python
df.shape
```

```python
df[(df['Department'] == 'Sales') & (df['DailyRate'] > 1000)] 
```

```python
df['DailyRate']
```

```python
def functions(x):
    return x/10

df['DailyRate'] = df['DailyRate'].map(functions)
df['DailyRate']
```

# Apply Method

```python
df[['DailyRate' , 'MonthlyRate']].apply(functions)
```

# ApplyMap

```python
df[['DailyRate' , 'MonthlyRate']].applymap(functions)
```
