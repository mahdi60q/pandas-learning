# Selecting Columns in Pandas

This notebook demonstrates different techniques for selecting columns from a Pandas DataFrame.

## Topics Covered

- Selecting individual columns
- Selecting multiple columns
- Selecting columns by data type
- Using `select_dtypes()`
- Working with numeric and non-numeric columns

## Goal

The goal is to learn how to efficiently select and work with specific columns in a DataFrame.

```python
import pandas as pd
```

```python
df = pd.read_csv('../data/Churn_Modelling.csv')
df.head()
```

```python
df.dtypes
```

```python
len(df.dtypes)
```

```python
len(df.select_dtypes(include='int64').columns)
```

```python
len(df.select_dtypes(exclude='int64').columns)
```
